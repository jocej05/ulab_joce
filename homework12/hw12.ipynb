## 2 Plotting
Make a scatter plot of **exoplanet Earth radius vs. equilibrium temperatures**. Include units, x/y axis labels, and a legend.

```python
import matplotlib.pyplot as plt

# Columns used 
x = df['pl_eqt']  # Equilibrium temperatures
y = df['pl_rade']  # Exoplanet Earth radii

# Creating scatter plot
plt.figure(figsize=(15, 10))
plt.scatter(x, y, color='pink', label='Exoplanets')

# Labels + title
plt.xlabel('Equilibrium Temperature (K)', fontsize=13)
plt.ylabel('Earth Radius (R)', fontsize=12)
plt.title('Exoplanet Earth Radius vs. Equilibrium Temperature', fontsize=15)

plt.legend()

# Print
plt.grid(True)
plt.show()
```

## 3 A "Cool" Planet
What planet has the **lowest** equilibrium temperature? Print the planet name and temperature with your `planet_data` variable.

```python
# Finding the planet with the lowest equilibrium temperature
lowest_temp_row = planet_data.loc[planet_data['pl_eqt'].idxmin()]

# Print results
print(f"Planet with the lowest equilibrium temperature:")
print(f"Name: {lowest_temp_row['pl_name']}, Temperature: {lowest_temp_row['pl_eqt']} K")
```

Make a new variable that only contains the information for that planet. \
\
*Hint: Use the `.iloc[[#]]` function. Replace # with the proper index.*

```python
# Finding the planet with the lowest equilibrium temperature
lowest_temp_index = planet_data[planet_data['pl_name'] == 'eps Ind A b'].index[0]

# Create a new variable
lowest_temp_planet = planet_data.iloc[[lowest_temp_index]]

# Print
print(lowest_temp_planet)
```
