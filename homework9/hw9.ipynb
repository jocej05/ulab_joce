# 2 Plotting a Function
Goal: Write a **function** that will plot $f(x) = x^2$ and $g(x) = \sqrt{x}$ on the same plot and return the $f(x)$ and $g(x)$ values as separate lists. Your plot needs to have the following:
- An appropriate figure size (`figsize=(10,8)` is recommended)
- Labeled x and y axis
- A title
- Different colors and linestyles for each function
- A legend
- Your domain should be $x \in [0, 10]$. *Hint: Use `np.linspace` to set a range of values for your functions.*

Your function should return:
- A list of $f(x)$ values
- A list of $g(x)$ values.

```python
import numpy as np
import matplotlib.pyplot as plt

def plot_functions():
    """
    Plots f(x) = x^2 and g(x) = sqrt(x) and includes labeled axes,
    title, and a legend outside the figure in the upper-right region.
    The function returns the calculated values of f(x) and g(x) 
    for the domain x in (0, 10).

    Returns:
    - fx_values: list of f(x) values where f(x) = x^2
    - gx_values: list of g(x) values where g(x) = sqrt(x)
    """
    # Using hint of np.linspace to set range of values
    x_values = np.linspace(0, 10, 100)
    
    # Calculate f(x) = x^2 and g(x) = sqrt(x)
    fx_values = x_values ** 2        # f(x) = x^2
    gx_values = np.sqrt(x_values)    # g(x) = sqrt(x)
    
    # Create the figure with specified dimensions
    plt.figure(figsize=(10, 8))
    
    # Plot f(x) with a label for the legend
    plt.plot(x_values, fx_values, color='pink', linestyle='-', label="f(x) = x^2")
    
    # Plot g(x) with a label for the legend
    plt.plot(x_values, gx_values, color='green', linestyle='--', label="g(x) = sqrt(x)")
    
    # Label the x-axis and y-axis
    plt.xlabel("X Values", fontsize=15)
    plt.ylabel("Function Values", fontsize=15)
    
    # Add a title to the plot
    plt.title("Plot of f(x) = x^2 and g(x) = sqrt(x)", fontsize=20)
    
    # Add a legend
    plt.legend()
    
    # Display the plot with all functions
    plt.grid(True)
    plt.show()
    
    # Return the lists of f(x) and g(x) values
    return fx_values.tolist(), gx_values.tolist()

# Call the function and store the returned fx and gx values
fx_values, gx_values = plot_functions()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def plot_functions():
    """
    Plots f(x) = x^2 and g(x) = sqrt(x) and includes labeled axes,
    title, and a legend outside the figure in the upper-right region.
    NEW: The function returns the calculated values of f(x) and g(x) 
    for the domain x in (0, 10), as well as the sums of each.

    Returns:
    - fx_values: list of f(x) values where f(x) = x^2
    - gx_values: list of g(x) values where g(x) = sqrt(x)
    NEW:
    
    - sum_fx: sum of all f(x) values
    - sum_gx: sum of all g(x) values
    """
    # Using np.linspace to set range of values
    x_values = np.linspace(0, 10, 100)
    
    # Calculate f(x) = x^2 and g(x) = sqrt(x)
    fx_values = x_values ** 2        # f(x) = x^2
    gx_values = np.sqrt(x_values)    # g(x) = sqrt(x)
    
    # Calculate the sum of f(x) and g(x) values
    sum_fx = sum(fx_values)
    sum_gx = sum(gx_values)
    
    # Create the figure with specified dimensions
    plt.figure(figsize=(10, 8))
    
    # Set the limits for the x and y axes
    plt.xlim(0, 15)       # Limit for x-axis
    plt.ylim(0, 100)      # Limit for y-axis

    # Plot f(x) with a label for the legend
    plt.plot(x_values, fx_values, color='pink', linestyle='-', label="f(x) = x^2")
    
    # Plot g(x) with a label for the legend
    plt.plot(x_values, gx_values, color='green', linestyle='--', label="g(x) = sqrt(x)")
    
    # Label the x-axis and y-axis
    plt.xlabel("X Values", fontsize=15)
    plt.ylabel("Function Values", fontsize=15)
    
    # Add a title to the plot
    plt.title("Plot of f(x) = x^2 and g(x) = sqrt(x)", fontsize=20)
    
    # Add a legend
    plt.legend() 
    
    # Display the plot with all functions
    plt.grid(True)
    plt.show()
    
    # Return the lists of f(x) and g(x) values, and their sums
    return fx_values.tolist(), gx_values.tolist(), sum_fx, sum_gx

# Call the function and store the returned values
fx_values, gx_values, sum_fx, sum_gx = plot_functions()
```

```python
import numpy as np
import matplotlib.pyplot as plt
#TAKEN FROM WEBSITE FOR PLOTTING A 3D FUNCTION: (Necessary library to generate a 3D image)
from matplotlib import cm
from matplotlib.ticker import LinearLocator

# Set up the X and Y axis values (from -10 to 10 with 100 points in between)
x = np.linspace(-10,10, 100)
y = np.linspace(-10, 10,100)

# Create a grid of X, Y points
X, Y = np.meshgrid(x, y)

# For a saddle surface 
# Define the Z values using a saddle function formula: Z = X^2 - Y^2
Z = X**2 - Y**2

# Make a 3D plot
fig = plt.figure()  # Create a new figure
ax = fig.add_subplot(111, projection='3d')  # Add 3D plot
ax.plot_surface(X, Y, Z, cmap='coolwarm')  # Plot the surface with color

# Add labels to the plot
ax.set_title("3D Saddle Surface")
ax.set_xlabel("X-axis")
ax.set_ylabel("Y-axis")
ax.set_zlabel("Z-axis (height)")

# Show the plot
plt.show()
```
